```python
def create_ann(w_img, h_img, num_hlayers, neurons_per_hlayer, num_results):
    len_layer1 = w_img * h_img
    g = {}
    #Notacion que quiero usar el nombre va a ser: l_n -> siendo l el numero de layer y n el numero de neurona en ese layer.
    for i in range(len_layer1):
        xd = "0_" + str(i)
        g[xd] = []
        
    for hlay in range(num_hlayers):
        for neuron in neurons_per_hlayer:
            xd = str(hlay+1) + "_" + str(neuron)
            g[xd] = []
            
    for i in range(num_results):
        xd = str(1+num_hlayers) + str(i)
        g[xd] = []
        
    
    # Las conexiones van a ser (i, w, f) -> I: inicial, W: peso, F:final.
    
    for layer in range(num_hlayers+2):
        if layer == 0:
            for neuron in range(w_img*h_img):
                xd1 = "0_" + str(neuron)
                for next_n in range(neurons_per_hlayer):
                    xd2="1_" + str(next_n)
                    g[xd1]=(xd1, 0, xd2)
                
        if layer != 0 and layer != (num_hlayers+1) and layer != num_hlayers:
            for neuron in range(neurons_per_hlayer):
                xd1 = str(layer) + "_" + str(neuron)
                for next_n in range(neurons_per_hlayer):
                    xd2 = str(layer+1) + "_" + str(next_n)
                    g[xd1] = (xd1, 0, xd2)
                    
                    
                    #Softmax -> mc
                    #sigmoide -> bin
                    #identidad / linear -> regression
```

Ok, so after some time trying to figure it out, I already understand the lineal algebra interpretation of neural networks, and my first approach is going to create some classes (OOP). In my analisys, I have identified 4 types of matrix, that i will explain rn:
- Input matrix ($I$)
- Weights matrixs ($W$)
- Hidden layer neurons matrixs ($H$)
- Output layer neurons matrix ($O$)

And, how i am just starting in this, we will start with just binary classification, so the output matrix will be of $1\times1$ (A constant).

And after we create those matrixs, we will start the fastforward, that will be litteraly this math operation:
$I \times W_1 = H_1 \times W_2 = \dots  = H_h \times W_{h+1} = S $.
Here I am excluding the rest of the fast forward steps like the bias and the activation function and all thats stuff.

```python
class Weights:
    def __init__(self, layer, input_neurons, output_neurons):
        self.type = 'Weights matrix'
        self.l = layer
        self.n = input_neurons
        self.m = output_neurons     
        self.temp_matrix = []
        
        for i in range(input_neurons):
            wn = []
            for j in range(output_neurons):
                weight = random.random()
                wn.append(weight)
            self.temp_matrix.append(wn)
            
        self.W = np.matrix(self.temp_matrix)
```
