```python
def create_ann(w_img, h_img, num_hlayers, neurons_per_hlayer, num_results):
    len_layer1 = w_img * h_img
    g = {}
    #Notacion que quiero usar el nombre va a ser: l_n -> siendo l el numero de layer y n el numero de neurona en ese layer.
    for i in range(len_layer1):
        xd = "0_" + str(i)
        g[xd] = []
        
    for hlay in range(num_hlayers):
        for neuron in neurons_per_hlayer:
            xd = str(hlay+1) + "_" + str(neuron)
            g[xd] = []
            
    for i in range(num_results):
        xd = str(1+num_hlayers) + str(i)
        g[xd] = []
        
    
    # Las conexiones van a ser (i, w, f) -> I: inicial, W: peso, F:final.
    
    for layer in range(num_hlayers+2):
        if layer == 0:
            for neuron in range(w_img*h_img):
                xd1 = "0_" + str(neuron)
                for next_n in range(neurons_per_hlayer):
                    xd2="1_" + str(next_n)
                    g[xd1]=(xd1, 0, xd2)
                
        if layer != 0 and layer != (num_hlayers+1) and layer != num_hlayers:
            for neuron in range(neurons_per_hlayer):
                xd1 = str(layer) + "_" + str(neuron)
                for next_n in range(neurons_per_hlayer):
                    xd2 = str(layer+1) + "_" + str(next_n)
                    g[xd1] = (xd1, 0, xd2)
                    
                    
                    #Softmax -> mc
                    #sigmoide -> bin
                    #identidad / linear -> regression
```
